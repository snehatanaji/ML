```python
def find_s(ex, target):
    h = ["φ"] * len(ex[0][0])
    for x, y in ex:
        if y == target:
            for i in range(len(x)):
                if h[i] == "φ":
                    h[i] = x[i]
                elif h[i] != x[i]:
                    h[i] = "?"
    return h

D = [
    (["Sunny","Hot","High","Weak"],"No"),
    (["Sunny","Hot","High","Strong"],"No"),
    (["Overcast","Hot","High","Weak"],"Yes"),
    (["Rainy","Mild","High","Weak"],"Yes"),
    (["Rainy","Cool","Normal","Weak"],"Yes"),
    (["Rainy","Cool","Normal","Strong"],"No")
]

print(find_s(D, "Yes"))
```

```python
from sklearn.datasets import load_wine
from sklearn.model_selection import train_test_split
from sklearn.ensemble import BaggingClassifier, AdaBoostClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score
X,y = load_wine(return_X_y=True)
Xtr,Xte,ytr,yte = train_test_split(X,y,test_size=.3,random_state=42)
bag = BaggingClassifier(DecisionTreeClassifier(),n_estimators=50,random_state=42)
boost = AdaBoostClassifier(DecisionTreeClassifier(max_depth=1),
n_estimators=50,random_state=42)
bag.fit(Xtr,ytr); boost.fit(Xtr,ytr)
print(accuracy_score(yte,bag.predict(Xte)))
print(accuracy_score(yte,boost.predict(Xte)))
```

```python
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt
X,_ = make_blobs(n_samples=200, centers=3, random_state=42)
m = KMeans(n_clusters=3, random_state=42)
m.fit(X)
plt.scatter(X[:,0],X[:,1],c=m.labels_)
plt.scatter(m.cluster_centers_[:,0],m.cluster_centers_[:,1],marker="X")
plt.show()
```

```python
!pip install minisom
import matplotlib.pyplot as plt
from minisom import MiniSom
from sklearn.datasets import load_iris
from sklearn.preprocessing import MinMaxScaler

iris = load_iris()
X = MinMaxScaler().fit_transform(iris.data)

som = MiniSom(7, 7, 4, sigma=1.0, learning_rate=0.5, random_seed=42)
som.random_weights_init(X)
som.train_random(X, 100)

plt.figure(figsize=(7, 7))
for x, t in zip(X, iris.target):
    w = som.winner(x)
    plt.text(w[0] + .5, w[1] + .5, str(t), ha="center", va="center",
             color=plt.cm.rainbow(t / 2), weight="bold")
plt.xlim(0, 7); plt.ylim(0, 7); plt.grid(True)
plt.title("SOM on Iris (labels 0, 1, 2)")
plt.show()
```

```python
!pip install pgmpy

from pgmpy.models import DiscreteBayesianNetwork
from pgmpy.factors.discrete import TabularCPD
from pgmpy.inference import VariableElimination
m = DiscreteBayesianNetwork([("Rain","Traffic"),("Accident","Traffic")])
r = TabularCPD("Rain",2,[[.7],[.3]])
a = TabularCPD("Accident",2,[[.9],[.1]])
t = TabularCPD("Traffic",2,
[[.9,.6,.7,.1],[.1,.4,.3,.9]],
evidence=["Rain","Accident"], evidence_card=[2,2])
m.add_cpds(r,a,t)
infer = VariableElimination(m)
print(infer.query(["Traffic"]))
```
